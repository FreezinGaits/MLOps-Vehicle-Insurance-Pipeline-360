```python
import pandas as pd
import pymongo

df = pd.read_csv('data.csv')
df.sample(5)
```

```python
# df should be converted into dict before we push it to mongodb

data = df.to_dict(orient='records')
# data
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

DB_NAME = os.getenv("DB_NAME") # logical database inside the cluster
COLLECTION_NAME = os.getenv("COLLECTION_NAME") # table-like container
CONNECTION_URL = os.getenv("CONNECTION_URL") # how Python finds & authenticates to MongoDB
```

```python
client = pymongo.MongoClient(CONNECTION_URL)
# This does not immediately connect.
# It creates a connection manager that:
# Handles pooling
# Handles authentication
# Handles failover internally
data_base = client[DB_NAME]
# This does not create the database yet
# It creates a reference to Proj1
collection = data_base[COLLECTION_NAME]
# Again, no creation yet
# Just a logical reference to Proj1-Data

# project -> database -> collection -> document -> record -> field -> value -> character -> byte -> bit -> electron -> quark -> string theory -> multiverse -> etc -> infinity -> beyond -> to be continued...
```

```python
# Uploading data to MongoDB
rec = collection.insert_many(data)
# This is the moment when everything becomes real.

# What happens internally:
# MongoDB creates database Proj1 (if not exists)
# Creates collection Proj1-Data (if not exists)
# Inserts each dictionary in data as a document
# Automatically adds _id field to each document
# Returns inserted IDs inside rec
```

```python
# Load back data from mongodb

df = pd.DataFrame(list(collection.find().limit(100)))
df.head(2)
```
